```python
def get_normalization(table, exclude=None):
    offset = table.nanmean()
    scale =  table.nanstd()
    
    print(offset)
    print(scale)
    
    for column in ak.fields(offset):
        if exclude is not None and column in exclude:
            offset[column] = [0.0]
            scale[column] = [1.0]
        if scale[column][0] == 0:  # Avoid NaN after divison
            scale[column] = [1.0]
    return offset, scale
```

```python
def load_dataset(path):
    with HDF5File(path, "r") as f:
        data   = f["data"]
        offset = f["offset"]
        scale  = f["scale"]
        for field in data.fields:
            data[field] = data[field] * scale[field] + offset[field]
        return data
```

```python
part_id = {
    "lepton": 1,
    "alepton": 2,
    "jet": 3,
    "top": 4,
    "atop": 5,
    "bot": 6, 
    "abot":7
}
num_types = len(part_id)

# Create a new dictionary with one-hot encodings
part_id_one_hot = {}

for key, value in part_id.items():
    one_hot = torch.zeros(num_types)  # Initialize a tensor of zeros with length equal to number of types
    one_hot[value - 1] = 1  # Set the corresponding index to 1 (value - 1 to convert to 0-based index)
    part_id_one_hot[key] = one_hot

# Display the one-hot encoded dictionary
for key, one_hot in part_id_one_hot.items():
    print(f"{key}: {one_hot}")

part_id_one_hot
```

```python
def process_particle_type(data_dict, pt_key, eta_key, phi_key, mass_key, particle_id, btag_value=0):
    pt = torch.tensor(data_dict[pt_key])
    eta = torch.tensor(data_dict[eta_key])
    phi = torch.tensor(data_dict[phi_key])
    mass = torch.tensor(data_dict[mass_key])
    
    # Stack the features
    features = torch.stack([pt, eta, phi, mass], dim=-1)
    
    # Create the one-hot encoded particle_id and btag tensors
    part_id_one_hot_tensor = torch.tensor(part_id_one_hot[particle_id], dtype=torch.float32).expand(features.size(0), -1)
    btag_tensor = torch.full((features.size(0), 1), btag_value, dtype=torch.float32)
    
    # Concatenate everything into a single tensor
    return torch.cat([features, part_id_one_hot_tensor, btag_tensor], dim=-1)

```

```python
# Process all particles in bulk
leptons = process_particle_type(combined_valid_data, "lep_pt", "lep_eta", "lep_phi", "lep_mass", 'lepton')
aleptons = process_particle_type(combined_valid_data, "alep_pt", "alep_eta", "alep_phi", "alep_mass", 'alepton')
top = process_particle_type(combined_valid_data, "top_pt", "top_eta", "top_phi", "top_mass", 'top')
atop = process_particle_type(combined_valid_data, "atop_pt", "atop_eta", "atop_phi", "atop_mass", 'atop')
bot = process_particle_type(combined_valid_data, "bot_pt", "bot_eta", "bot_phi", "bot_mass", 'bot', btag_value=1)
abot = process_particle_type(combined_valid_data, "abot_pt", "abot_eta", "abot_phi", "abot_mass", 'abot', btag_value=1)

```

```python
# Process jets, which are variable-length, in bulk and pad them later
jet_list = []
for index in range(len(combined_valid_data["lep_pt"])):
    jet_pt = torch.tensor(combined_valid_data["jet_pt"][index])
    jet_eta = torch.tensor(combined_valid_data["jet_eta"][index])
    jet_phi = torch.tensor(combined_valid_data["jet_phi"][index])
    jet_mass = torch.tensor(combined_valid_data["jet_mass"][index])
    jet_btag = torch.tensor(combined_valid_data["jet_btag"][index])
    
    jets = torch.stack([jet_pt, jet_eta, jet_phi, jet_mass], dim=-1)
    
    part_id_one_hot_tensor = torch.tensor(part_id_one_hot['jet'], dtype=torch.float32).expand(jets.size(0), -1)
    
    # Concatenate jets with particle_id and btag
    jets = torch.cat([jets, part_id_one_hot_tensor, jet_btag.unsqueeze(-1)], dim=-1)
    jet_list.append(jets)
```

```python
input_events = []
for i in range(len(leptons)):
    # Combine lepton, alepton, and jets for the event
    event_particles = torch.cat([leptons[i].unsqueeze(0), aleptons[i].unsqueeze(0), jet_list[i]], dim=0)
    
    # Pad with zeros to reach 256 particles
    padding = torch.zeros(256 - event_particles.size(0), event_particles.size(1))
    event_particles = torch.cat([event_particles, padding], dim=0)
    
    input_events.append(event_particles)

input_tensor = torch.stack(input_events)
```

```python
# Combine targets for each event
target_events = []
for i in range(len(top)):
    target_event = torch.stack([top[i], atop[i], bot[i], abot[i]], dim=0)
    target_events.append(target_event)

target_tensor = torch.stack(target_events)
```

```python
# Now input_tensor and target_tensor are ready for your DataLoader
print(f"Input Tensor Shape: {input_tensor.shape}")
print(f"Target Tensor Shape: {target_tensor.shape}")
```

```python
event_weights = torch.tensor(combined_valid_data["weight"])
```

```python
dataset = TensorDataset(input_tensor, target_tensor, event_weights)

# Step 2: Define the directory and filename
directory = f'/nfs/dust/cms/user/coradine/DATA_SPECIAL_DM_SM_100_cpt/DM_{int(a)}_SM_{int(100-a)}'
filename = 'validatedata.pt'

# Step 3: Create the directory if it doesn't exist
os.makedirs(directory, exist_ok=True)

# Step 4: Save the Dataset
torch.save(dataset, os.path.join(directory, filename))
```
